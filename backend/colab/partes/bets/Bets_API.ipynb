```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter
import requests

#endpoint
url = "https://6aa1d1d4ccb3db9689a5cff3.mockapi.io/endividamento-mf-2026"
response = requests.get(url)
df = pd.DataFrame(response.json())


df['Data_Ord'] = pd.to_datetime(df['Mes'], format='%m/%Y')
df = df.sort_values('Data_Ord').reset_index(drop=True)

fig, ax1 = plt.subplots(figsize=(12, 6))

#função para formatar os valores do eixo y em Bilhões (R$)
def formatar_bilhoes(x, pos):
    return f'R$ {x / 1e9:.1f} Bi'

color_verde = 'tab:green'
ax1.set_xlabel('Mês', fontsize=12, labelpad=10)
ax1.set_ylabel('Depósito Total (R$)', color=color_verde, fontsize=12)
line1 = ax1.plot(df['Mes'], df['Deposito-mes'], color=color_verde, marker='o', linewidth=2.5, label='Depósito (R$)')
ax1.tick_params(axis='y', labelcolor=color_verde)

ax1.yaxis.set_major_formatter(FuncFormatter(formatar_bilhoes))
plt.xticks(rotation=45)

#função para formatar valores do eixo y secundário em milhões (apostadores)
def formatar_milhoes(x, pos):
    return f'{x / 1e6:.1f} Mi'

ax2 = ax1.twinx()
color_vermelho = 'tab:red'
ax2.set_ylabel('Número de Apostadores', color=color_vermelho, fontsize=12)
line2 = ax2.plot(df['Mes'], df['Numero-apostadores'], color=color_vermelho, marker='s', linestyle='--', linewidth=2.5, label='Nº Apostadores')
ax2.tick_params(axis='y', labelcolor=color_vermelho)
ax2.grid(False)

ax2.yaxis.set_major_formatter(FuncFormatter(formatar_milhoes))

lines = line1 + line2
labels = [line.get_label() for line in lines]
ax1.legend(lines, labels, loc='upper left')

plt.title('Evolução de Apostas no Brasil (Janeiro 2025 - Junho 2026)', fontsize=14, fontweight='bold', pad=15)
fig.tight_layout()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import requests

#endpoint
url = "https://6aa1d1d4ccb3db9689a5cff3.mockapi.io/endividamento-mf-2026"
response = requests.get(url)
df = pd.DataFrame(response.json())

df['Data_Ord'] = pd.to_datetime(df['Mes'], format='%m/%Y')
df = df.sort_values('Data_Ord').reset_index(drop=True)

plt.style.use('ggplot')
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 10), sharex=True)

#função para formatar valores do 1 grafico
def formatar_bilhoes(x, pos):
    return f'R$ {x / 1e9:.1f} Bi'

color_verde = 'tab:green'
ax1.bar(df['Mes'], df['Deposito-mes'], color=color_verde, alpha=0.85, width=0.6)
ax1.set_ylabel('Depósito Total (R$)', color=color_verde, fontsize=12, fontweight='bold')
ax1.tick_params(axis='y', labelcolor=color_verde)
ax1.tick_params(axis='x', labelbottom=True, labelrotation=45)
ax1.set_title('Evolução de Depósitos e Apostadores no Brasil (Jan/2025 - Jun/2026)', fontsize=14, fontweight='bold', pad=15)

ax1.yaxis.set_major_formatter(plt.FuncFormatter(formatar_bilhoes))

#função para formatar valores do 2 grafico
def formatar_milhoes(x, pos):
    return f'{x / 1e6:.1f} Mi'

color_vermelho = 'tab:red'
ax2.bar(df['Mes'], df['Numero-apostadores'], color=color_vermelho, alpha=0.85, width=0.6)
ax2.set_ylabel('Número de Apostadores', color=color_vermelho, fontsize=12, fontweight='bold')
ax2.tick_params(axis='y', labelcolor=color_vermelho)
ax2.set_xlabel('Mês', fontsize=12, labelpad=10)

ax2.yaxis.set_major_formatter(plt.FuncFormatter(formatar_milhoes))

plt.xticks(rotation=45)

fig.tight_layout()

plt.show()
```
